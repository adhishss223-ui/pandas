```python

# STEP 1: Import libraries

import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report


# STEP 2: Load dataset

df = pd.read_csv("iris.csv")

print(df.head())




X = df.drop("Species", axis=1)
y = df["Species"]




numeric_features = X.select_dtypes(include=["int64", "float64"]).columns

categorical_features = X.select_dtypes(include=["object"]).columns


# STEP 5: Create numerical preprocessing

numeric_pipeline = make_pipeline(
    SimpleImputer(strategy="mean"),
    StandardScaler()
)


# STEP 6: Create categorical preprocessing

categorical_pipeline = make_pipeline(
    SimpleImputer(strategy="most_frequent"),
    OneHotEncoder(handle_unknown="ignore")
)


# STEP 7: Combine preprocessing

preprocessor = ColumnTransformer([
    ("num", numeric_pipeline, numeric_features),
    ("cat", categorical_pipeline, categorical_features)
])


#

model = make_pipeline(
    preprocessor,
    RandomForestClassifier(random_state=42)
)


# STEP 9: Split training and testing data

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.3,
    random_state=42,
    stratify=y
)


# STEP 10: Train model

model.fit(X_train, y_train)

print("Training completed!")


# STEP 11: Predict

y_pred = model.predict(X_test)

print("Predicted values:")
print(y_pred)


# STEP 12: Check accuracy

accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)


# STEP 13: Classification report

print(classification_report(y_test, y_pred, zero_division=0))


# STEP 14: New prediction

new_flower = pd.DataFrame(
    [[5.1, 3.5, 1.4, 0.2]],
    columns=X.columns
)

result = model.predict(new_flower)

print("Predicted flower:", result[0])
```
